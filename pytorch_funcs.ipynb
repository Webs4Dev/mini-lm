```python
import torch
```

#### torch.multinomial 

```python
# each probability points to the index of that probability in tensor
probability = torch.tensor([0.1,0.7,0.2]) # 0->10%,1->70%,2->20% 
samples = torch.multinomial(probability,num_samples=10,replacement=True)
print(samples)
```

#### torch.cat

```python
tensor = torch.tensor([1,2,3,4]) 
output = torch.cat((tensor,torch.tensor([5])),dim=0) # concatenate two tensors
print(output)
```

#### torch.tril

```python
lt = torch.tril(torch.ones(4,4)) # lower triangular matrix
print(lt)
```

#### torch.triu

```python
ut = torch.triu(torch.ones(4,4)) # upper triangular matrix
print(ut)
```

#### torch.masked_fill

```python
mf = torch.zeros(5,5).masked_fill(torch.tril(torch.ones(5,5))==0,float('-inf')) # masking values
print(mf)
```

#### torch.exp

```python
exp = torch.exp(torch.triu(torch.ones(3,3))) # e = 2.71 so basically every value becomes e^value 
print(torch.triu(torch.ones(3,3)))
print(exp)
```

#### input.transpose

```python
input = torch.ones(4,2,3)
transpose = input.transpose(0,2)
print(transpose.shape)
```

#### torch.stack

```python
tensor1 = torch.tensor([1,2,3])
tensor2 = torch.tensor([1,2,3])
tensor3 = torch.tensor([1,2,3])

stacked = torch.stack([tensor1,tensor2,tensor3])
print(stacked)
```

#### nn.linear

```python
import torch.nn as nn # Neural Networl
sample = torch.tensor([10,10,10],dtype=torch.float)
linear = nn.Linear(3,1,bias=False) # nn.Linear(no. of input layers=,no=. of output layers=, bias=) 
print(linear(sample))
```

#### softmax function


```python
import torch.nn.functional as F
tensor = torch.tensor([1,2,3],dtype=torch.float)
softmax = F.softmax(tensor,dim=0) # formula = e^i/(e^i+e^i+1....e^n) 
print(softmax)
```

#### nn.Embeddings

```python
import torch
import torch.nn as nn

# Number of unique tokens (IDs: 0–79)
vocab_size = 80

# Size of vector for each token
embedding_dim = 6

# Create embedding matrix of shape (80,6)
embedding = nn.Embedding(vocab_size, embedding_dim)

# Generate 10 random token IDs
input_indices = torch.randint(0,vocab_size,size=(10,))

# Convert token IDs → embedding vectors
output = embedding(input_indices)

print("Shape:", output.shape)
```

#### Matrix Multiplication

```python
m1 = torch.LongTensor(torch.randint(1,10,size=(2,3)))
m2 = torch.LongTensor(torch.randint(1,10,size=(3,2)))
matmul = torch.matmul(m1,m2) # same as m1 @ m2
print(matmul)
```
